```python
import matplotlib.pyplot as plt
import numpy as np

# 배경
fig, ax = plt.subplots(figsize=(8, 8))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

# 불꽃 중심
cx, cy = 0, 0

# 불꽃의 기본 선 만들기
for angle in np.linspace(0, 2*np.pi, 24, endpoint=False):
    length = np.random.uniform(1.5, 3.5)

    x = [cx, cx + length * np.cos(angle)]
    y = [cy, cy + length * np.sin(angle)]

    ax.plot(x, y, linewidth=1.5)

# 중심점
ax.scatter(cx, cy, s=20)

# 화면 설정
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 화면 설정
fig, ax = plt.subplots(figsize=(8, 8))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

# 불꽃 중심
cx, cy = 0, 0

# 불꽃의 선
for angle in np.linspace(0, 2*np.pi, 36, endpoint=False):

    # 길이와 흔들림
    length = np.random.uniform(1.5, 3.5)

    t = np.linspace(0, 1, 30)

    # 약간 휘어진 선
    x = cx + length * t * np.cos(angle)
    y = cy + length * t * np.sin(angle)

    # 붓의 흔들림처럼 약간의 랜덤 변화
    x += np.random.normal(0, 0.03, len(x))
    y += np.random.normal(0, 0.03, len(y))

    # 선을 여러 번 겹쳐 번지는 느낌
    ax.plot(
        x, y,
        linewidth=np.random.uniform(0.5, 2.0),
        alpha=np.random.uniform(0.4, 0.9)
    )

# 불꽃 주변의 작은 입자
for i in range(120):

    angle = np.random.uniform(0, 2*np.pi)
    distance = np.random.uniform(1.0, 3.8)

    x = cx + distance * np.cos(angle)
    y = cy + distance * np.sin(angle)

    ax.scatter(
        x, y,
        s=np.random.uniform(2, 15),
        alpha=np.random.uniform(0.2, 0.8)
    )

# 중심부
ax.scatter(cx, cy, s=30)

# 화면 설정
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, IntSlider, FloatSlider

def draw_fireworks(
    firework_count=3,
    size=2.5,
    particle_count=80
):

    fig, ax = plt.subplots(figsize=(8, 8))

    fig.patch.set_facecolor("black")
    ax.set_facecolor("black")

    # 여러 개의 불꽃 생성
    for _ in range(firework_count):

        # 불꽃 위치
        cx = np.random.uniform(-3, 3)
        cy = np.random.uniform(-3, 3)

        # 불꽃 선
        for angle in np.linspace(0, 2*np.pi, 30, endpoint=False):

            length = np.random.uniform(
                size * 0.5,
                size
            )

            t = np.linspace(0, 1, 25)

            x = cx + length * t * np.cos(angle)
            y = cy + length * t * np.sin(angle)

            # 약간의 흔들림
            x += np.random.normal(0, 0.03, len(x))
            y += np.random.normal(0, 0.03, len(y))

            ax.plot(
                x, y,
                linewidth=np.random.uniform(0.5, 1.8),
                alpha=np.random.uniform(0.4, 0.9)
            )

        # 주변 입자
        for _ in range(particle_count):

            angle = np.random.uniform(0, 2*np.pi)
            distance = np.random.uniform(
                size * 0.4,
                size * 1.2
            )

            x = cx + distance * np.cos(angle)
            y = cy + distance * np.sin(angle)

            ax.scatter(
                x, y,
                s=np.random.uniform(2, 12),
                alpha=np.random.uniform(0.2, 0.8)
            )

        # 중심
        ax.scatter(cx, cy, s=25)

    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_aspect("equal")
    ax.axis("off")

    plt.show()


interact(
    draw_fireworks,

    firework_count=IntSlider(
        min=1,
        max=8,
        step=1,
        value=3,
        description="불꽃 개수"
    ),

    size=FloatSlider(
        min=1,
        max=4,
        step=0.5,
        value=2.5,
        description="불꽃 크기"
    ),

    particle_count=IntSlider(
        min=20,
        max=150,
        step=10,
        value=80,
        description="입자 개수"
    )
)
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(10, 7))

fig.patch.set_facecolor("black")
ax.set_facecolor("black")


def draw_firework(cx, cy, size, alpha, rays=32):

    # 불꽃의 중심
    for angle in np.linspace(0, 2*np.pi, rays, endpoint=False):

        length = np.random.uniform(size * 0.7, size)

        t = np.linspace(0, 1, 35)

        # 약간 휘어진 불꽃
        curve = np.sin(t * np.pi) * np.random.uniform(-0.15, 0.15)

        x = cx + length * t * np.cos(angle)
        y = cy + length * t * np.sin(angle)

        x += curve * np.sin(angle)
        y += curve * np.cos(angle)

        # 불꽃 선
        ax.plot(
            x, y,
            linewidth=np.random.uniform(0.5, 1.8),
            alpha=alpha
        )

    # 주변 입자
    for _ in range(int(size * 35)):

        angle = np.random.uniform(0, 2*np.pi)
        distance = np.random.uniform(size * 0.5, size * 1.2)

        x = cx + distance * np.cos(angle)
        y = cy + distance * np.sin(angle)

        ax.scatter(
            x, y,
            s=np.random.uniform(2, 12),
            alpha=alpha * np.random.uniform(0.3, 0.8)
        )

    # 중심
    ax.scatter(
        cx, cy,
        s=size * 8,
        alpha=alpha
    )


# --------------------------------
# 뒤쪽 불꽃
# --------------------------------

for _ in range(8):

    cx = np.random.uniform(-7, 7)
    cy = np.random.uniform(-2, 4)

    draw_firework(
        cx,
        cy,
        size=np.random.uniform(0.7, 1.3),
        alpha=np.random.uniform(0.15, 0.35)
    )


# --------------------------------
# 중간 불꽃
# --------------------------------

for _ in range(5):

    cx = np.random.uniform(-6, 6)
    cy = np.random.uniform(-2, 4)

    draw_firework(
        cx,
        cy,
        size=np.random.uniform(1.3, 2.0),
        alpha=np.random.uniform(0.4, 0.65)
    )


# --------------------------------
# 앞쪽 큰 불꽃
# --------------------------------

for _ in range(3):

    cx = np.random.uniform(-5, 5)
    cy = np.random.uniform(-1, 4)

    draw_firework(
        cx,
        cy,
        size=np.random.uniform(2.2, 3.2),
        alpha=np.random.uniform(0.7, 1.0)
    )


# 화면 설정
ax.set_xlim(-9, 9)
ax.set_ylim(-4, 7)

ax.set_aspect("equal")
ax.axis("off")

plt.show()
```
